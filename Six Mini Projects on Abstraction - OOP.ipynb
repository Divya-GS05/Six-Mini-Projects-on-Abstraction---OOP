```python
#Task 1 - Payment Gateway
from abc import ABC, abstractmethod
class Payment(ABC):
  def __init__(self,amount):
    self.amount = amount
  @abstractmethod
  def pay(self):
    pass
  def refund(self):
    pass
class UPIPayment(Payment):
  def pay(self):
    return f"UPI payment of {self.amount} is successful"
  def refund(self,cash):
    return f"UPI refund of {cash} is successful"
class CardPayment(Payment):
  def pay(self):
    return f"Card payment of {self.amount} is successful"
  def refund(self,netbanking):
    return f"Card refund of {netbanking} is successful"

upi = UPIPayment("5000")
print(upi.pay())
print(upi.refund("2000"))
card = CardPayment("6000")
print(card.pay())
print(card.refund("1000"))


```

```python
#Task 2 - Notification service
from abc import ABC, abstractmethod
class Notification(ABC):
  def __init__(self,message):
    self.message = message
  @abstractmethod
  def notify(self):
    pass

class SMSNotification(Notification):
  def notify(self):
    return f"SMS sent: {self.message}"

class EmailNotification(Notification):
  def notify(self):
    return f"Email sent: {self.message}"

class PushNotification(Notification):
  def notify(self):
    return f"Push Notification sent: {self.message}"


def checkout(notification):
  print(notification.notify())

checkout(SMSNotification("Hi"))
checkout(EmailNotification("Congrats"))
checkout(PushNotification("Offer"))
```

```python
#Task 3 - Database Repository
from abc import ABC, abstractmethod
class DatabaseRepository(ABC):
  @abstractmethod
  def add(self,name):
    pass
  def get(self,user_id):
    pass
  def delete(self,id):
    pass

class PostgreSQL(DatabaseRepository):
  def add(self,name):
    return f"Adding {name} to PostgreSQL"
  def get(self,user_id):
    return f"Getting {user_id} from PostgreSQL"
  def delete(self,id):
    return f"Deleting {id} from PostgreSQL"

class MongoDB(DatabaseRepository):
  def add(self,name):
    return f"Adding {name} to MongoDB"
  def get(self,user_id):
    return f"Getting {user_id} from MongoDB"
  def delete(self,id):
    return f"Deleting {id} from MongoDB"

postgre = PostgreSQL();
print(postgre.add("Priya"))
print(postgre.get(1234))

mongodb = MongoDB();
print(mongodb.add("Divya"))
print(mongodb.get(4567))
```

```python
#Task 4 - File Storage System
from abc import ABC, abstractmethod
class FileStorage(ABC):
  @abstractmethod
  def upload(self,file):
    pass
  def download(self,file):
    pass
  def delete(self,file):
    pass

class LocalStorage(FileStorage):
  def upload(self,file):
    return f"Uploading {file} to LocalStorage"
  def download(self,file):
    return f"Downloading {file} to LocalStorage"
  def delete(self,file):
    return f"Deleting {file} from LocalStorage"

class AWS_S3(FileStorage):
  def upload(self,file):
    return f"\nUploading {file} to AWS S3"
  def download(self,file):
    return f"Downloading {file} to AWS S3"
  def delete(self,file):
    return f"Deleting {file} from AWS S3"

class Google_Cloud_Storage(FileStorage):
  def upload(self,file):
    return f"\nUploading {file} to Google Cloud Storage"
  def download(self,file):
    return f"Downloading {file} to Google Cloud Storage"
  def delete(self,file):
    return f"Deleting {file} from Google Cloud Storage"


class Azure_Blob_Storage(FileStorage):
  def upload(self,file):
    return f"\nUploading {file} to Azure Blob Storage"
  def download(self,file):
    return f"Downloading {file} to Azure Blob Storage"
  def delete(self,file):
    return f"Deleting {file} from Azure Blob Storage"


local = LocalStorage()
print(local.upload("file.txt"))
print(local.download("data.txt"))
print(local.delete("old file.txt"))

aws = AWS_S3()
print(aws.upload("resume.txt"))
print(aws.download("details.txt"))
print(aws.delete("old resume.txt"))

google = Google_Cloud_Storage()
print(google.upload("python.txt"))
print(google.download("range.txt"))
print(google.delete("old python.txt"))

azure = Azure_Blob_Storage()
print(azure.upload("java.txt"))
print(azure.download("array.txt"))
print(azure.delete("old java.txt"))
```

```python
#Task 6 - Report Generator
from abc import ABC, abstractmethod
import json

class ReportGenerator(ABC):
    @abstractmethod
    def export(self, data):
        pass

class CSVReport(ReportGenerator):
    def export(self, data):
        # Return CSV header as a string
        return ",".join(data)

class JSONReport(ReportGenerator):
    def export(self, data):
        return json.dumps(data)

class ExcelReport(ReportGenerator):
    def export(self, data):
        return f"Excel format exported with headers: {data}"

class PDFReport(ReportGenerator):
    def export(self, data):
        return f"PDF format exported with headers: {data}"

# Match the exact user output requirement
headers = ['id', 'name', 'city']
csv_rep = ExcelReport()
print(csv_rep.export(headers))
```
